```python
#read csv file
df = pd.read_csv('/content/drive/MyDrive/Data_PHC_New.csv')
df.head()
```

```python
# --- governance indicators ---
# Indicator 1: Legitimate Standing (max score = 4)
df['Indicator1_Legitimate_Standing_of_health_facilities'] = (
    df['Facility_has_a_management_unit_FMC__FMB_constituted_via_either_a_gazette_notice_or_an_appointment_letter_outlining_a_list_of_board_members'] +
    df['Facility_management_unit_minutes_available'] +
    df['Facility_has_an_up_to_date_licence_from_the_medical_practitioners_and_dentists_council']
) / 4 * 100

# Indicator 2: PCN Functionality (max score = 5)
df['Indicator2_Functionality_of_facility_within_PCN'] = (
    df['Facility_notification__contracting_letter_available_linking_it_to_a_gazetted_PCN'] +
    df['Facility_has_filed_briefs_from_MDT_inreach_visit_available'] +
    df['Facility_has_filed_action_plans_from_support_supervision_visit_available']
) / 5 * 100
```

```python
# --- Information Systems Indicators ---
# Indicator 3: Data Use (max = 8)
df['Indicator3_Data_use_for_programatic_decision_makinig'] = (
    df['Facility_has_HIS_data_review_meeting_minutes_available'] +
    df['Facility_has_mortality_audit_meeting_minutes_available'] +
    df['Facility_outpatient_workload_is_reported_regularly_eg_reported_every_month_in_the_past_three_months'] +
    df['Facility_summarises_the_leading_causes_of_disease_from_outpatient_attendance_registers'] +
    df['Facility_catchment_population_is_up_to_date_2025_on_HISKenya']
) / 8 * 100

# Indicator 4: EMR Functionality (assume 8 components)
df['Indicator4_Functionality_of_EMR_systems'] = (
    df['Facility_has_an_electronic_medical_record_EMR_system_installed'] +
    df['Facility_EMR_system_is_working_and_in_use_routinely'] +
    df['EMR_system_operates_in_all_major_facility_departments_and_is_not_restricted_to_HIVTB_program_only'] +
    df['EMR_can_capture_basic_demographic_and_clinical_health_information'] +
    df['EMR_provides_clinical_decision_support_functions_including_alerts_prompts_and_reminders'] +
    df['EMR_allows_for_order_entry_and_prescribing'] +
    df['EMR_possesses_functionality_for_encryption_and_maintenance_of_confidentiality_of_patient_data'] +
    df['EMR_provides_for_electronic_information_exchange_via_agreed_protocols']
) / 8 * 100
```

```python
# Indicator 5: Infrastructure (max = 6)
df['Indicator5_Infrastructure_for_service_delivery'] = (
    df['Facility_has_working_internet_connection_on_the_day_of_visit'] +
    df['Facility_has_electricity__power_running_on_the_day_of_visit'] +
    df['Facility_has_clean_tapped_water_available_on_the_day_of_visit'] +
    df['Facility_outpatient_department_has_usable__working_toilets'] +
    df['Facility_has_at_least_3_visible_waste_segregation_bins_with_appropriate_bin_liners'] +
    df['Facility_has_a_plan_for_client_referral_and_a_linked_referral_facility']
) / 6 * 100

# Indicator 6: Equipment (max = 7)
df['Indicator6_Structural_components_for_quality_PHC'] = (
    df['Paediatric_weighing_scale_at_general_outpatient_clinic_is_working_and_well_zeroed'] +
    df['Blood_pressure_monitoring_machine_at_the_general_outpatient_clinic_is_working'] +
    df['Pulse_oximeter_at_the_general_outpatient_clinic_is_working'] +
    df['Emergency__Resuscitation_tray_is_accessible_at_casualty_outpatient_department'] +
    df['Flow_meter_for_oxygen_therapy_available_and_working_at_maternity'] +
    df['Neonatal_resuscitation_bag_and_mask_Ambubag_available_at_maternity'] +
    df['Delivery_couch_available_at_maternity_section']
) / 7 * 100
```

```python
# --- Commodity availability ---
#medicines
df['Medicines'] = (
    df['Adrenaline_Inj_1mgml'] +
    df['Amoxycillin_Dispersible_Tablets_250mg_10s'] +
    df['Chlorhexidine_gel_71pc_tube_20g'] +
    df['Magnesium_Sulphate_Injection_50pc_10ml'] +
    df['Tranexamic_Acid_Injection_1g_10ml'] +
    df['Oxytocin_10IU__or__Heat_Stable_Cabetocin_injection'] +
    df['ORS_Zinc_copack']
) / 7 * 100

df['Medical_Supplies'] = (
    df['Cord_Clamp'] +
    df['Foley_catheter_18G_30mL_2way'] +
    df['Maternity_pad_dim_26X9X1'] +
    df['Safety_box_WHO_specification_5L'] +
    df['Nasal_prongs_for_oxygen_delivery_paediatric_size']
) / 5 * 100

df['Diagnostic_Supplies'] = (
    df['Glucose_test_strips'] +
    df['Haemoglobin_cuvettes__measuring_kit'] +
    df['HIV_rapid_test'] +
    df['Syphilis_rapid_test'] +
    df['Lugols_reagent_Iodine_35pc_for_Cervical_Cancer_screening'] +
    df['Urinalysis_10_parameter_strips']
) / 6 * 100

df['Commodity_availability'] = (
    df['Medicines'] +
    df['Medical_Supplies'] +
    df['Diagnostic_Supplies']
) / 3 * 100
```

```python
# --- Human Resources Ratios ---
# Mapping of hospital levels to HRH norms
hrh_norms_map = {
    'Level 2': 5,
    'Level 3a': 27,
    'Level 3b': 27,
    'Level 4': 120
}

# Assign norms based on the Level column
df['Norms'] = df['Level'].map(hrh_norms_map)

#staff count
df['Total_core_staff'] = (df['Medical_officers'] + df['Clinical_officers'] + df['BSc_nursing_cadre'] + df['Enrolled_nurses'] + df['Laboratory_specialists'] + df['Laboratory_technologists'] + df['Registered_nurses'] + df['Pharmacists'] + df['Pharmaceutical_technologists'])
df['Absent_staff'] = (df['Medical_officers1'] + df['Clinical_officers1'] + df['BSc_nursing_cadre1'] + df['Enrolled_nurses1'] + df['Laboratory_specialists1'] + df['Laboratory_technologists1'] + df['Registered_nurses1'] + df['Pharmacists1'] + df['Pharmaceutical_technologists1'])

# --- Absence Ratio ---
df['Absence_ratio'] = 100 * (df['Absent_staff'] / df['Total_core_staff'])

# --- staff to norm ratio ---
df['staff_to_norm_ratio'] = 100 * (df['Total_core_staff'] / df['Norms'])

# --- workload ratio ---
df['Workload_ratio'] = 10000 * (df['Total_core_staff'] / df['Total_monthly_outpatient_workload_previous_month'])

# --- Percentage of deployed staff vs HRH Norms ----
df['Staff_present_percent'] = 100 * (df['Total_core_staff'] / df['Norms'])

# --- Gap to target HRH NOrms ---
df['Staff_absent_percent'] = 100 * (df['Norms'] - df['Total_core_staff'])
```

```python
# Save the updated DataFrame to a new Excel file
df.to_csv('data_phc_pm_new.csv', index=False)
```

```python
#--- summarise all the calculations by subcounties ----
# List of indicator columns to summarize
indicator_cols = [
    'Indicator1_Legitimate_Standing_of_health_facilities',
    'Indicator2_Functionality_of_facility_within_PCN',
    'Indicator3_Data_use_for_programatic_decision_makinig',
    'Indicator4_Functionality_of_EMR_systems',
    'Indicator5_Infrastructure_for_service_delivery',
    'Indicator6_Structural_components_for_quality_PHC',
    'Medicines', 'Medical_Supplies', 'Diagnostic_Supplies',
    'Commodity_availability',
    'Absence_ratio',
    'staff_to_norm_ratio',
    'Workload_ratio',
    'Staff_present_percent',
    'Staff_absent_percent'
  ]

# Group by Sub County and calculate the mean for each indicator
summary_by_subcounty = df.groupby('Sub_County')[indicator_cols].mean().reset_index()

# Save the summary to CSV
summary_by_subcounty.to_csv("PHC_Indicator_Summary_By_SubCounty.csv", index=False)

# Display the summary (optional)
print(summary_by_subcounty)

```

```python
#join the data to geojson file
import geopandas as gpd

# Load your indicator data
indicators = pd.read_csv("/content/data_phc_pm_new.csv")

# Load your GeoJSON (must have matching sub-county names or codes)
geo = gpd.read_file("/content/drive/MyDrive/Nakuru_TransNzoiaSubCounty.geojson")

# Preview GeoJSON columns to find the matching key
print(geo.columns)
```

```python
# Make sure both datasets use the same name format
indicators['Sub_County'] = indicators['Sub_County'].str.strip().str.lower()
geo['Sub_County'] = geo['name'].str.strip().str.lower()

# Merge indicator scores into GeoJSON
geo_merged = geo.merge(indicators, on='Sub_County', how='left')

# Save merged file
geo_merged.to_file("PHC_Indicators_By_SubCounty.geojson", driver='GeoJSON')

```
